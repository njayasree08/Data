**Insight DataSet Analysis**

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(6,4))
sns.countplot(x='gender',data=df)
plt.title('Gender Analysis')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(6,4))
sns.countplot(x='Partner',data=df)
plt.title('Partern Analysis')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(6,4))
sns.countplot(x='Dependents',data=df)
plt.title('Dependent Analysis')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,4))
sns.barplot(x='PaymentMethod',y='Churn',data=df)
plt.title('PaymentMethod  Analysis')
plt.xlabel('PaymentMethod')
plt.ylabel('Count')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,4))
sns.barplot(x='Contract',y='Churn',data=df)
plt.title('Contract  Analysis')
plt.xlabel('Contract')
plt.ylabel('Count')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,4))
sns.barplot(x='InternetService',y='Churn',data=df)
plt.title('InternetService Analysis')
plt.xlabel('InternetService')
plt.ylabel('Count')
plt.show()
```

```python
# Monthly distribution
plt.figure(figsize=(10, 6))
sns.histplot(df['MonthlyCharges'], bins=30, kde=True)
plt.title('Distribution of MonthlyCharges')
plt.xlabel('MonthlyCharges')
plt.ylabel('Frequency')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(4, 4))
sns.heatmap(df.corr(numeric_only=True), fmt='.2f')
plt.title('Correlation matrix for numerical features')
plt.show()
```

```python
import pandas as pd


plt.figure(figsize=(10, 6))
sns.lineplot(x='tenure', y='TotalCharges', data=df , marker='o')
plt.title('Total Charges vs. Tenure')
plt.xlabel('Tenure (months)')
plt.ylabel('Total Charges')
plt.show()
```
